# Lab 14 - Using Multimodal LLMs for Visual Question Answering
This notebook uses Hugging Face Transformers with the `dandelin/vilt-b32-finetuned-vqa` model.

```python
from transformers import pipeline
from PIL import Image

image = Image.open(r'C:\Users\licha\OneDrive\Desktop\Gen-AI\output\playwright\lab14_sample_cats.jpg')
vqa_pipe = pipeline('visual-question-answering', model='dandelin/vilt-b32-finetuned-vqa')
print('VQA model loaded successfully')
```

```python
questions = [
    'How many cats are there?',
    'What are the cats sitting on?',
    'What color is the couch?',
]
for question in questions:
    result = vqa_pipe(image=image, question=question, top_k=3)
    print(question)
    print(result[0])
    print()
```
